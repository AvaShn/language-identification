# Text Evaluation and Error Analysis

This notebook inspects actual held-out prediction errors from the CV-selected raw-text pipeline.

```python
from language_identification.config import load_config
from language_identification.common.io import read_json
config = load_config()
selection = read_json(config.artifacts_dir / 'text' / 'classification' / 'metrics' / 'selection.json')
selection
```

```python
import pandas as pd
predictions = pd.read_csv(config.artifacts_dir / 'text' / 'classification' / 'predictions' / f"{selection['best_model_key']}.csv")
errors = predictions.loc[predictions.true_language != predictions.predicted_language]
errors.groupby(['true_language', 'predicted_language']).size().sort_values(ascending=False).head(10)
```
